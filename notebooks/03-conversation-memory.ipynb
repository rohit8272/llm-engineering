```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="qwen3.8-27b",  # Replace with an available model ID (right now, qwen3.8-27b is available in experiential labs)
    api_key=os.environ["EXPLABS_API_KEY"],
    base_url="https://api.experientiallabs.ai/v1",
)
```

```python
history = []

while True:
    query =  input("User: ")
    history.append({"role": "user", "content": query})
    print("User: ", query)

    if query.lower() in ["exit", "quit", "bye"]:
        break

    res = llm.invoke(history)

    history.append({"role": "ai", "content": res.content})
    print("AI: ", res.content, "\n")
```
