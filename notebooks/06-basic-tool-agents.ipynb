```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from langchain_openai import ChatOpenAI


llm = ChatOpenAI(
    model="qwen3.8-27b",  # Replace with an available model ID
    api_key=os.environ["EXPLABS_API_KEY"],
    base_url="https://api.experientiallabs.ai/v1",
)
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """ 
        this function add two numbers and return the sum of two input numbers 

        Args
            1. first number
            2. second number
    """
    return a+b
```

```python
@tool
def multiply_numbers(a:int, b:int):
    """ 
        this function multiply two numbers and return the product of two input numbers 

        Args
            1. first number
            2. second number
    """
    return a*b
```

```python
from langchain.agents import create_agent

agent = create_agent(
                model=llm ,
                tools=[add_numbers, multiply_numbers],
                system_prompt="You are a math teacher, and always use tool for calculation."
            )
```

```python
response = agent.invoke({"messages" : [{"role" : "user", "content" : "what is 2 + 5 * 20 ?"}]})
```
