```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="qwen3.8-27b",  # Replace with an available model ID (right now, qwen3.8-27b is available in experiential labs)
    api_key=os.environ["EXPLABS_API_KEY"],
    base_url="https://api.experientiallabs.ai/v1",
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator who translates English to {language}."},
    {"role": "user", "content": "{query}"},
])

prompts  = prompt_template.format_messages(language="Hindi", query="Hello, how are you?")

response = llm.invoke(prompts)
print(response.content)
```

```python
from langchain_core.output_parsers import StrOutputParser

out = StrOutputParser()
```

```python
def ConvertToUpperCase(text: str):
    return text.upper()
```

```python
chains = prompt_template | llm | out | ConvertToUpperCase

res = chains.invoke({"language": "Hinglish", "query": "Hello, how are you?"})
print(res)
```
